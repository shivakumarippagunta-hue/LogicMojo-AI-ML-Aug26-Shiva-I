## Problem: 4

Create a 10 x 10 arrays of zeros and then "frame" it with a border of ones.

```python
# Write your code here
arr = np.zeros((10, 10), dtype=int)

arr[0, :] = 1       # Top border
arr[-1, :] = 1      # Bottom border
arr[:, 0] = 1       # Left border
arr[:, -1] = 1      # Right border

print(arr)
```

## Problem: 5

Create an 8 x 8 array with a checkerboard pattern of zeros and ones.

```python
# Write your code here

arr = np.zeros((8, 8), dtype=int)

arr[1::2, ::2] = 1
arr[::2, 1::2] = 1

print(arr)

```

## Problem: 6

Create an array $x$ with 100 points in the interval $[-1,1]$ (the points must be uniformly spaced). Then create another array $y$. 

Where $y = x^2 sin(x^2) + x$. 

Try to plot $y$ vs $x$ using matplotlib.

```python
# Write your code here
import numpy as np
import matplotlib.pyplot as plt

# Create 100 uniformly spaced points between -1 and 1
x = np.linspace(-1, 1, 100)

# Calculate y
y = x**2 * np.sin(x**2) + x

# Plot y vs x
plt.plot(x, y)

# Add labels and title
plt.xlabel("x")
plt.ylabel("y")
plt.title("y = x² sin(x²) + x")

# Display the graph
plt.show()

```

## Problem: 7 
Create a function which creates an $n×n$ array with $(i , j)$ -entry equal to $i+ j$

```python
# Write your code here

def create_array(n):
    arr = np.zeros((n, n), dtype=int)

    for i in range(n):
        for j in range(n):
            arr[i, j] = i + j

    return arr


n = int(input("Enter n: "))

print(create_array(n))

```

# Problem: 8
Evaluate $cos$ and $sin$ on the interval $[0,1]$  with 1000 points and then stack the results into a tall array with rows being the $(cos(x),sin(x))$ entries.

```python
# Write your code here
# Create 1000 uniformly spaced points from 0 to 1
x = np.linspace(0, 1, 1000)

# Calculate cos(x) and sin(x)
cos_x = np.cos(x)
sin_x = np.sin(x)

# Stack them as columns
result = np.column_stack((cos_x, sin_x))

print(result)
print("Shape:", result.shape)
```

# Problem: 9

Create a random $3×5$ array using the `np.random.rand(3 ,5)` function and compute: the sum of all the entries, the sum of the rows, and the sum of the columns.

```python
# Write your code here


# Create a random 3 × 5 array
arr = np.random.rand(3, 5)

print("Array:")
print(arr)

# Sum of all entries
total_sum = np.sum(arr)

# Sum of each row
row_sum = np.sum(arr, axis=1)

# Sum of each column
column_sum = np.sum(arr, axis=0)

print("\nSum of all entries:", total_sum)
print("Sum of each row:", row_sum)
print("Sum of each column:", column_sum)

```

## Problem: 10

Create a random $5×5$ array using the function `np.random.rand(5, 5)`. Then sort the rows according to the second column.

```python
# Write your code here
# Create a random 5 × 5 array
arr = np.random.rand(5, 5)

print("Original array:")
print(arr)

# Sort rows according to the second column
sorted_arr = arr[arr[:, 1].argsort()]

print("\nSorted array:")
print(sorted_arr)
```
