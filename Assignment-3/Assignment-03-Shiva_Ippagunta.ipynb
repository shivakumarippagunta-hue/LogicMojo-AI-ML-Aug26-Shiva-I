#### 6. What are the five most common County names in the U.S.?

```python
# Write your code here
import re

def clean_county_name(name):
    # Remove common suffixes like 'County', 'Parish', 'Borough', 'City and Borough', 'Census Area'
    name = re.sub(r' County$| Parish$| Borough$| City and Borough$| Census Area$| city$', '', name, flags=re.IGNORECASE)
    return name.strip()

# Apply the cleaning function to the 'County' column
pop['CleanedCountyName'] = pop['County'].apply(clean_county_name)

# Count the occurrences of each cleaned county name
common_county_names = pop['CleanedCountyName'].value_counts()

# Get the five most common names
top_5_common_county_names = common_county_names.head(5)

print("The five most common county names are:")
print(top_5_common_county_names)

```

#### 7. What are the top 5 most populated Counties according to the 2010 Census?

```python
# Write your code here
top_5_counties = pop.sort_values(by='2010Census', ascending=False).head(5)
print(top_5_counties[['County', 'State', '2010Census']])

```

#### 8. What are the top 5 most populated States according to the 2010 Census?

```python
# Write your code here
state_population = pop.groupby('State')['2010Census'].sum()
top_5_states = state_population.sort_values(ascending=False).head(5)
print(top_5_states)

```

#### 9. How many Counties have 2010 populations greater than 1 million?

```python
# Write your code here
counties_over_1m = pop[pop['2010Census'] > 1000000]
num_counties_over_1m = len(counties_over_1m)
print(f"There are {num_counties_over_1m} counties with a 2010 population greater than 1 million.")

```

#### 10. How many Counties don't have the word "County" in their name?

```python
# Write your code here
counties_without_word_county = pop[~pop['County'].str.contains('County', case=False, na=False)]
num_counties_without_word_county = len(counties_without_word_county)
print(f"There are {num_counties_without_word_county} counties that do not have the word 'County' in their name.")

```

#### 11. Add a column that calculates the percent change between the 2010 Census and the 2017 Population Estimate

```python
# Write your code here
pop['PercentChange'] = ((pop['2017PopEstimate'] - pop['2010Census']) / pop['2010Census']) * 100
print(pop.head())

```

#### 12. What States have the highest estimated percent change between the 2010 Census and the 2017 Population Estimate?

**NOTE:** This will take several lines of code, as it requires a recalculation of Percent Change.

```python
# Write your code here
state_percent_change = pop.groupby('State')['PercentChange'].mean()
highest_percent_change_states = state_percent_change.sort_values(ascending=False).head(5)
print(highest_percent_change_states)

```
